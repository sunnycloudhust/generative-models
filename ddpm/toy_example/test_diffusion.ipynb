```python
import torch
import numpy as np
import matplotlib.pyplot as plt

from dataset import dataloader, show_tensor_image
from model import forward_diffusion_sample, T

sample_batch = next(iter(dataloader))
sample_image = sample_batch[1]

num_images = 8
steps = np.linspace(0, T - 1, num_images, dtype=int)

plt.figure(figsize=(18, 3))
plt.subplot(1, num_images + 1, 1)
show_tensor_image(sample_image.unsqueeze(0))
plt.title("Original")

for idx, t_idx in enumerate(steps, start=2):
    t = torch.tensor([t_idx], dtype=torch.long)
    noisy_image, _ = forward_diffusion_sample(sample_image.unsqueeze(0), t)
    plt.subplot(1, num_images + 1, idx)
    show_tensor_image(noisy_image)
    plt.title(f"t={t_idx}")

plt.tight_layout()
plt.show()
```
